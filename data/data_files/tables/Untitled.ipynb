```python
import os
import pandas as pd

folder_path = r"testdata"
output_file = os.path.join(folder_path, "combined.xlsx")

excel_files = [
    f for f in os.listdir(folder_path)
    if f.endswith(".xlsx") and f != "combined.xlsx"
]

with pd.ExcelWriter(output_file, engine="openpyxl") as writer:
    sheet_written = False

    for file in excel_files:
        file_path = os.path.join(folder_path, file)

        try:
            df = pd.read_excel(file_path)
            sheet_name = os.path.splitext(file)[0][:31]  # Excel limit
            df.to_excel(writer, sheet_name=sheet_name, index=False)
            sheet_written = True

        except Exception as e:
            print(f"Skipping {file}: {e}")

    if not sheet_written:
        # Ensure at least one visible sheet exists
        pd.DataFrame({"info": ["No valid Excel files found"]}).to_excel(
            writer, sheet_name="Info", index=False
        )
```

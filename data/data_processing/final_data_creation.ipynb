```python
import pandas as pd
import numpy as np



# Define date range
start_date = pd.to_datetime("2022-09-01").date()
end_date   = pd.to_datetime("2025-09-30").date()

# Generate all months in the period
months = pd.date_range(start=start_date, end=end_date, freq="MS").date

# Function to assign random dates per customer
def assign_random_dates(customer_df):
    n = len(customer_df)
    selected_dates = []

    if n >= len(months):
        # guarantee one random date per month
        for m in months:
            month_start = pd.to_datetime(m).date()
            month_end = (pd.to_datetime(m) + pd.offsets.MonthEnd(0)).date()
            random_day = np.random.randint((month_end - month_start).days + 1)
            selected_dates.append(month_start + pd.Timedelta(days=random_day))
        
        # remaining random dates in full range
        remaining = n - len(months)
        random_days = np.random.randint(
            0, (end_date - start_date).days + 1, remaining
        )
        random_dates = [start_date + pd.Timedelta(days=int(d)) for d in random_days]
        all_dates = np.concatenate([selected_dates, random_dates])
    else:
        # if fewer transactions than months, just random sample
        random_days = np.random.randint(
            0, (end_date - start_date).days + 1, n
        )
        all_dates = [start_date + pd.Timedelta(days=int(d)) for d in random_days]

    np.random.shuffle(all_dates)
    return pd.Series(all_dates, index=customer_df.index)

# Apply per customer
df_data["random_date_2"] = (
    df_data.groupby("customer_id")
      .apply(assign_random_dates)
      .reset_index(level=0, drop=True)
)


```

```python
# import pandas as pd
# import numpy as np

# Your existing DataFrame
# df has columns: country, amount

# Exchange rate map, including Nigeria
rates = {
    "Canada": 0.725632,
    "UK": 1.348348,
    "Brazil": 0.187908,
    "USA": 1.0,
    "Japan": 0.006759,
    "Australia": 0.659350,
    "Russia": 0.011976,
    "Germany": 1.174553,  # assuming EUR → USD
    "France": 1.174553,   # if you have both
    "Singapore": 0.779155,
    "Mexico": 0.054396,
    "Nigeria": 0.000669   # 1 NGN to USD
}

def convert_to_usd(row):
    c = row["country"]
    rate = rates.get(c)
    if rate is None:
        # handle missing country: you can set to NaN or some default
        return np.nan
    return row["amount"] * rate

df_data["amount_usd"] = df_data.apply(convert_to_usd, axis=1)
```
